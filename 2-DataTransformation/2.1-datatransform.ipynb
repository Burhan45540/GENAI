```python
from langchain_community.document_loaders import TextLoader
textload=TextLoader("genai.txt").load()
textload
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

textsplitter=RecursiveCharacterTextSplitter(chunk_size=50,chunk_overlap=20)

textsplitter.split_documents(textload)
```

```python
html_string = """
<!DOCTYPE html>
<html>
<head>
    <title>Artificial Intelligence</title>
</head>
<body>
    <h1>Welcome to Artificial Intelligence</h1>
    <p>Artificial Intelligence (AI) enables machines to perform tasks 
    that typically require human intelligence.</p>

    <h2>Applications of AI</h2>
    <ul>
        <li>Natural Language Processing</li>
        <li>Machine Learning</li>
        <li>Computer Vision</li>
        <li>Generative AI</li>
    </ul>

    <h2>Generative AI</h2>
    <p>Generative AI can create new content, including text, images, 
    audio, and code. Large Language Models (LLMs) are widely used 
    to build intelligent chatbots and question-answering systems.</p>

    <a href="https://www.openai.com">Learn more about AI</a>
</body>
</html>
"""

header_split=[
    ("h1","Header1"),
    ("h2","para")
]
    
```

```python
from langchain_text_splitters import HTMLHeaderTextSplitter

htmlsplitter=HTMLHeaderTextSplitter(headers_to_split_on=header_split)

html=htmlsplitter.split_text(html_string)
```

```python
import json
import requests

json_data=requests.get("https://raw.githubusercontent.com/cjbatin/Swift4-Decoding-JSON-Using-Codable/master/WeatherForecast/StubData/londonWeather.json").json()
json_data
```

```python
from langchain_text_splitters import RecursiveJsonSplitter

jsonsplitter=RecursiveJsonSplitter(max_chunk_size=20)

json_chunk=jsonsplitter.split_json(json_data)
```
