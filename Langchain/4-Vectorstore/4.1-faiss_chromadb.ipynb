```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader("edu.txt").load()
loader
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(chunk_size=150,chunk_overlap=30)
docs=splitter.split_documents(loader)
docs
```

```python
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
embeddings=OllamaEmbeddings(model="nomic-embed-text")
db=FAISS.from_documents(docs,embeddings)
```

```python
db
```

```python
query1="Why are hard work and consistency both important for achieving success? Explain with an example from your own life."
query2="Imagine you are learning a difficult skill, but you are not getting good results even after several weeks. What would you do to stay motivated and continue learning?"
```

```python
db.similarity_search(query1)
```

```python
db.similarity_search(query2)
```

```python
db.similarity_search_with_score(query1)
```

```python
db.similarity_search_with_score(query2)
```

```python
db.save_local("faiss_index")
```

```python
new_db=FAISS.load_local("faiss_index",embeddings=embeddings,allow_dangerous_deserialization=True)
```

```python
db1=new_db.similarity_search_with_score(query1)
db1
```
