```python
from langchain_community.document_loaders import WebBaseLoader
import bs4

webloader=WebBaseLoader(web_path=("https://en.wikipedia.org/wiki/Rohit_Sharma"),
                        bs_kwargs=dict(parse_only=bs4.SoupStrainer(
                            class_=("vector-page-toolbar")
                        ))
                        )
```

```python
webloader.load()
```
